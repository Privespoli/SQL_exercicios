###1. Ejercicio para ver que temperatura hace cada día de la semana com grafico de linea

```python
import pandas as pd
import matplotlib.pyplot as plt
datos = {
    'Día': ['Lun', 'Mar', 'Mie', 'Jue', 'Vie', 'Sab', 'Dom'],
    'Temp': [22, 24, 23, 28, 26, 30, 31]
}

df = pd.DataFrame(datos)
df.head()
```

```python
plt.plot(df['Día'],df['Temp'], marker="D", color="aqua")
plt.title("Temperatura ")
plt.xlabel("Día de la semana ")
plt.ylabel("Grados Celsius")
plt.show()
```

###2. Queremos graficar cuantos cafés tomaste esta semana:

```python
dias = ['Lunes', 'Martes', 'Miércoles', 'Jueves', 'Viernes']
cafes = [2, 4, 3, 1, 5]

plt.plot(dias, cafes, marker="o", color="#ff69b4")
plt.title("Cafés tomados por día")
plt.xlabel("Día de la semana")
plt.ylabel("Cantidad de cafés")
plt.show()
```

###3. Crea un gráfico de barras comparando la cantidad de inventario de 4 frutas: Manzanas (50), Bananas (30), Cerezas (45) y Peras (20).

```python
frutas = ['Manzanas', 'Bananas', 'Cerezas', 'Peras']
cantidades = [50, 30, 45, 20]
plt.bar(frutas, cantidades, color =["#ff471a", "#cca300", "#330000", "#5bbd5b"], label ="Inventario actual", edgecolor ="black", linewidth = 2, hatch =".....")
plt.title("Inventario de frutas", fontsize="14", fontweight="bold")
plt.xlabel("Frutas")
plt.ylabel("Cantidad en inventario")
plt.legend()
plt.show()
```

###4. Relacion estatura-peso: usa un grafico de dispersion (scatter) con 5 puntos de datos inventados para ver si hay una relación visual entre la altura de una persona y su peso

```python
estaturas = [160, 170, 180, 155, 190, 200]
pesos = [60, 72, 85, 50, 95, 75]
plt.scatter(estaturas, pesos, color="purple", s=100, marker="o")
plt.title("Relación Estatura-Peso")
plt.xlabel("Estatura (cm)")
plt.ylabel("Peso (kg)")
plt.grid(True)
plt.show()
```

###5. Distribución de Exámenes: Imagina que tienes una lista de 20 notas de alumnos (del 1 al 10). Crea un histograma para ver qué notas son las más frecuentes.

```python
notas = [1, 5, 8, 9, 7, 10, 4, 6, 8, 8, 9, 5, 3, 7, 8, 10, 2, 9, 8, 7]

plt.hist(notas, bins=range(1, 12), align='left', rwidth=0.8, color='pink', edgecolor='black')

plt.title("Distribución de Notas de Alumnos", fontsize=14, fontweight="bold")
plt.xlabel("Notas")
plt.ylabel("Frecuencia")
plt.xticks(range(1, 11))
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### 6. Quieres analizar cómo gastas tu tiempo en un día de 24 horas. Este es un caso de uso clásico para un gráfico de sectores.

```python
import matplotlib.pyplot as plt

actividades = ['Dormir', 'Comer', 'Trabajo/Estudio', 'Ocio', 'Deporte']
horas = [8, 2, 8, 4, 2]
desfase = [0, 0, 0, 0.12, 0] # Desfasamos un poco el ocio

# Gama de colores pasteles elegantes
colores_pastel = ['#ffb3ba', '#ffdfba', '#ffffba', '#bffcc6', '#c9c9ff']

# Creamos la figura con un tamaño agradable
plt.figure(figsize=(7, 7))

# Graficamos con sombra, bordes limpios y estilo redondeado
plt.pie(
    horas,
    labels=actividades,
    colors=colores_pastel,
    explode=desfase,
    autopct='%1.1f%%',
    startangle=140,
    shadow=True,              # Efecto 3D de sombra
    wedgeprops={'edgecolor': 'white', 'linewidth': 1.5, 'antialiased': True} # Bordes limpios
)

plt.title("Distribución de Tiempo en Horas", fontsize=14, fontweight="bold", pad=20)
plt.show()
```
